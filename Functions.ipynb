```python
def AVERAGE():
    try:
        n1=eval(input('enter num1: '))
        n2=eval(input('enter num2: '))
        n3=eval(input('enter num3: '))
        avg=round((n1+n2+n3)/3,2)
        print(f'avg are: {avg}')
    except Exception as e:
        print(e)
AVG()
```

```python
AVERAGE()
```
